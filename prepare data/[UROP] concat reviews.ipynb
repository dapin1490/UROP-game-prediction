```python
import pandas as pd

lowSeller_path = "/content/drive/MyDrive/Colab Notebooks/data/gameWReview_02.csv"
topSeller_path = "/content/drive/MyDrive/Colab Notebooks/data/topSellGameWReview.csv"

lowSellers = pd.read_csv(lowSeller_path)
topSellers = pd.read_csv(topSeller_path)

lowSellers.drop(["Unnamed: 0", "Unnamed: 0.1"], axis=1, inplace=True)
topSellers.drop(["Unnamed: 0", "Unnamed: 0.1"], axis=1, inplace=True)

lowSellers["total_review"] = lowSellers["positive review"] + lowSellers["negative review"]
topSellers["total_review"] = topSellers["positive review"] + topSellers["negative review"]

lowSellers["is_top_seller"] = 0

topName = list(topSellers["title"])

for i in range(len(lowSellers["appid"])):
    gName = lowSellers["title"][i]
    if topName.count(gName) > 0:
        lowSellers["appid"][i] = None

totalSellers = pd.concat([lowSellers, topSellers], axis=0)
totalSellers.sort_values(by='appid', inplace=True)
totalSellers.dropna(subset=['appid'], inplace=True)
totalSellers.drop_duplicates()
totalSellers.reset_index(drop=True, inplace=True)

for i in range(len(totalSellers["appid"])):
    rvtext = totalSellers["all positive"][i]
    rvcnt = totalSellers["total_review"][i]
    if str(rvtext).lower().find("positive") >= 0 and rvcnt >= 500:
        totalSellers["is_top_seller"][i] = 1

totalSellers.to_csv("/content/drive/MyDrive/Colab Notebooks/data/totalSellersWReview.csv")

totalSellers.info()
```
